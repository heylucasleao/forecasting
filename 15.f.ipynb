5. Onda complexa de fourier

```python
def get_fourier_wave(x, k):
    x = np.asarray(x)

    N = len(x)
    t = np.arange(N)

    wave = np.exp(-2j * np.pi * k * t / N)

    return wave
```

```python
get_fourier_wave(x, k=1)
```

```python
def fourier_coefficient(x, k):
    x = np.asarray(x, dtype=float)

    N = len(x)
    t = np.arange(N)

    wave = np.exp(-2j * np.pi * k * t / N)

    return np.sum(x * wave)
```

```python
def fourier_coefficients(x):
    x = np.asarray(x, dtype=float)

    N = len(x)
    t = np.arange(N)
    k_values = np.arange(N // 2 + 1)

    coefficients = []

    for k in k_values:
        wave = np.exp(-2j * np.pi * k * t / N)
        X_k = np.sum(x * wave)

        coefficients.append(X_k)

    return np.array(coefficients)
```

```python
coefficients = fourier_coefficients(x)
```

```python
magnitude = np.abs(coefficients) 
power = magnitude ** 2
```

```python
power
```

```python
real = coefficients.real
imag = coefficients.imag

power = real**2 + imag**2
```
