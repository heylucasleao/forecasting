2. Posição temporal (dado que a variação seja 1)

```python
def get_time_index(x):
    x = np.asarray(x)
    return np.arange(len(x))
```

```python
t = get_time_index(x)
t
```

3. Coletar os indices das frequencias

```python
def get_frequency_indices(x):
    x = np.asarray(x)
    N = len(x)

    return np.arange(N // 2 + 1)
```

```python
k = get_frequency_indices(x)
k
```

4. Transformar cada k em frequência

```python
frequencies  = k / len(x)
```
